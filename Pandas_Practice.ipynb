```python
# Pandas Data Handling Notebook
import pandas as pd

df = pd.read_csv("titanic.csv")

print("--- First 5 rows ---")
print(df.head())

# 2. Indexing & Slicing
print("--- Selecting specific columns ---")
print(df[['Survived', 'Sex', 'Age']].head())

print("--- Using loc (label-based) ---")
print(df.loc[0:5, ['Survived','Age','Pclass']])

print("--- Using iloc (position-based) ---")
print(df.iloc[0:5, 0:3])

# 3. Data Cleaning
print("--- Checking null values ---")
print(df.isnull().sum())

# Fill missing ages with mean
df['Age'] = df['Age'].fillna(df['Age'].mean())
# Drop rows with NA in 'Embarked'
df = df.dropna(subset=['Embarked'])
# Rename column
df = df.rename(columns={'SibSp':'Siblings_Spouses'})
# Drop column 'Cabin' (too many nulls)
df = df.drop(columns=['Cabin'])

# 4. Aggregation & Grouping
print("--- Groupby Example (Survival rate by Sex) ---")
print(df.groupby('Sex')['Survived'].mean())

print("--- Pivot Table (Survival by Sex and Class) ---")
print(df.pivot_table(values='Survived', index='Sex', columns='Pclass', aggfunc='mean'))

# 5. Merging & Joining
extra_info = pd.DataFrame({
    'Pclass': [1,2,3],
    'fare_discount': [10, 5, 2]
})

merged = pd.merge(df, extra_info, on='Pclass', how='left')
print("--- Merged Data Example ---")
print(merged[['Pclass','Fare','fare_discount']].head())

# 6. Exporting
# Save cleaned dataset
df.to_csv('titanic_clean.csv', index=False)
df.to_excel('titanic_clean.xlsx', index=False)

# Export summary statistics
summary = df.describe()
summary.to_csv('titanic_summary.csv')

print("--- Exported cleaned dataset and summary to CSV/Excel ---")
```
