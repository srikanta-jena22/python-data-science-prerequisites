```python
# Seaborn Visualization Notebook

import seaborn as sns
import matplotlib.pyplot as plt

# Load example datasets
iris = sns.load_dataset("iris")
titanic = sns.load_dataset("titanic")

# 1. Distribution Plots
# Histogram
sns.histplot(iris["sepal_length"], bins=20, kde=True)
plt.title("Histogram with KDE")
plt.show()

# KDE Plot
sns.kdeplot(iris["sepal_width"], fill=True)
plt.title("KDE Plot Example")
plt.show()

# 2. Categorical Plots
# Countplot
sns.countplot(x="class", data=titanic)
plt.title("Count of Passengers by Class")
plt.show()

# Barplot
sns.barplot(x="class", y="fare", data=titanic, errorbar=None)
plt.title("Average Fare by Class")
plt.show()

# Boxplot
sns.boxplot(x="class", y="age", data=titanic)
plt.title("Age Distribution by Class")
plt.show()

# Violinplot
sns.violinplot(x="sex", y="age", data=titanic)
plt.title("Age Distribution by Gender")
plt.show()

# 3. Relational Plots
# Scatterplot
sns.scatterplot(x="sepal_length", y="petal_length", hue="species", data=iris)
plt.title("Scatterplot of Iris Dataset")
plt.show()

# Lineplot
sns.lineplot(x="age", y="fare", data=titanic, errorbar=None)
plt.title("Fare vs Age")
plt.show()

# 4. Heatmap
# Correlation heatmap of Titanic numeric features
corr = titanic.corr(numeric_only=True)
sns.heatmap(corr, annot=True, cmap="coolwarm")
plt.title("Titanic Correlation Heatmap")
plt.show()

# 5. Pairplot & Jointplot
sns.pairplot(iris, hue="species")
plt.suptitle("Iris Pairplot", y=1.02)
plt.show()

sns.jointplot(x="age", y="fare", data=titanic, kind="scatter")
plt.suptitle("Age vs Fare Jointplot", y=1.02)
plt.show()

# Task: Titanic Dataset Visualizations
# Survival vs Age
sns.histplot(data=titanic, x="age", hue="survived", multiple="stack")
plt.title("Survival by Age")
plt.show()

# Survival vs Class
sns.barplot(x="class", y="survived", data=titanic)
plt.title("Survival Rate by Class")
plt.show()

# Survival vs Gender
sns.barplot(x="sex", y="survived", data=titanic)
plt.title("Survival Rate by Gender")
plt.show()
```
